```python
import pandas as pd
```

```python
# 1. Load Data
raw = pd.read_csv("teaching_prices_messy.csv")
clean = raw.copy()
```

```python
# 2. Remove Exact Duplicates
clean = clean.drop_duplicates()
```

```python
#3. Standardize Categorical Strings
clean["item"] = clean["item"].str.strip().str.title()
```

```python
# 4. Correct Mislabeled Item (Fixing the 23 vs 25 count anomaly)
# On 2026-02-05 in City 1, Oil B was recorded twice and Staple A was missing. 
# Index 40 (price 108.5) matches Staple A's trend, so we fix the label.
clean.loc[40, "item"] = "Staple A"
```

```python
# 5. Clean Numeric Formatting and Convert Data Types
# Removes commas from prices (e.g., '1,250') and casts to float
clean["price"] = pd.to_numeric(clean["price"].astype(str).str.replace(",", ""), errors="coerce")
```

```python
# 6. Convert Dates
clean["date"] = pd.to_datetime(clean["date"])
```

```python
# --- BEFORE AND AFTER COMPARISON TABLE ---
print("--- BEFORE CLEANING ---")
print(f"Total Rows: {raw.shape[0]}")
print(f"Missing Values in Price: {raw['price'].isna().sum()}")
print(f"Item Counts (Raw):\n{raw['item'].value_counts().head(3)}\n")
print("="*40)
print("\n--- AFTER CLEANING ---")
print(f"Total Rows: {clean.shape[0]}")
print(f"Missing Values in Price: {clean['price'].isna().sum()}")
print(f"\nFinal Item Counts (Should ALL be 24 - 12 weeks * 2 cities):")
print(clean['item'].value_counts())
print("\nData Types:")
print(clean.dtypes)
```
